```python
import matplotlib.pyplot as plt

attrition_counts = df["Attrition"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(attrition_counts.index, attrition_counts.values)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

overtime_attrition
```

```python
overtime_attrition["Yes"].plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

department_attrition
```

```python
department_attrition["Yes"].sort_values(ascending=False).plot(
    kind="bar"
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
jobrole_attrition = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

jobrole_attrition.sort_values("Yes", ascending=False)
```

```python
jobrole_attrition["Yes"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Attrition Rate by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Key Business Insights

1. **Overall Attrition**
   - Out of 1,470 employees, 237 employees left the organization.
   - The overall attrition rate was **16.12%**.

2. **Overtime and Attrition**
   - Employees working overtime had an attrition rate of **30.53%**.
   - Employees not working overtime had an attrition rate of only **10.44%**.
   - This indicates a strong association between overtime and employee attrition.

3. **Department-wise Attrition**
   - The **Sales department** recorded the highest attrition rate at **20.63%**.
   - Human Resources had an attrition rate of **19.05%**.
   - Research & Development had the lowest departmental attrition rate at **13.84%**.

4. **Job Role Analysis**
   - **Sales Representatives** had the highest attrition rate at **39.76%**.
   - Laboratory Technicians had an attrition rate of **23.94%**.
   - Research Directors had the lowest attrition rate at **2.50%**.

5. **Business Travel**
   - Employees who travelled frequently had an attrition rate of **24.91%**.
   - Employees who travelled rarely had an attrition rate of **14.96%**.
   - Non-travelling employees had the lowest rate at **8.00%**.

6. **Age and Attrition**
   - Employees who left had an average age of **33.61 years**.
   - Employees who stayed had an average age of **37.56 years**.

7. **Income and Attrition**
   - Employees who left had an average monthly income of **4,787.09**.
   - Employees who stayed had an average monthly income of **6,832.74**.

## Business Recommendations

- Review workload and overtime policies, particularly for employees who regularly work overtime.
- Investigate retention challenges within the Sales department, especially among Sales Representatives.
- Review the workload and support provided to employees who travel frequently.
- Examine compensation and career-growth opportunities for lower-income employees.
- Develop targeted retention strategies for younger employees and high-attrition job roles.
